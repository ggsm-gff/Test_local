```python
import pandas as pd
from sqlalchemy import create_engine, text
import pyodbc as odbc
import json

connection_parameters = '''{
"andromeda" : 
    "DRIVER={ODBC Driver 17 for SQL Server};    SERVER=andromeda;    DATABASE=MIS_RIESGOS;    Trusted_Connection=yes;"
}'''
connection_string = json.loads(connection_parameters)['andromeda']
engine = create_engine('mssql+pyodbc:///?odbc_connect='+ connection_string)

with engine.connect() as conn:
    df = pd.read_sql_query('select getdate() as fecha_actual', con=conn)

df.head()
```
