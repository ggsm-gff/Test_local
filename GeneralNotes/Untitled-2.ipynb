```python
from thefuzz import process
import numpy as np
choices = ['american','italian','asian']

def get_clean_type(dirty_type, choices, threshold=80):
    if dirty_type is None:
        return np.NaN
    if len(dirty_type) == 0:
        return np.NaN
    if len(choices) == 0:
        return np.NaN

    result = process.extractOne(dirty_type, choices=choices)
    #print(result)
    if result[1]<=threshold:
        return np.NaN
    else:
        return result[0]
    

df['type_clean'] = df['type'].apply(get_clean_type, choices=choices)

df
```
