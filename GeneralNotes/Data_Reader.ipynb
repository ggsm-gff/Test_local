```python
from sqlalchemy import create_engine
import pandas as pd
import json

#Reads the conf file with the definition to connect to andromeda.
#config_path = 'C:\\Users\\HN34064\\Desktop\\PPR\\Dev\\src\\config.json'
config_path = 'config.json'
with open(config_path,'r',encoding='utf-8') as file:
    config = json.load(file)

connection_string = config['andromeda']

#Reads the SQL query to execute it later.
#sql_path= 'C:\\Users\\HN34064\\Desktop\\PPR\\Dev\\src\\SQL-Models\\Balances_Hist.sql'
sql_path= 'SQL-Models/Balances_Hist.sql'
with open(sql_path,'r',encoding='utf-8') as file:
    query = file.read()

#Creates a sqlalchemy engine to interact with the DB.
engine = create_engine('mssql+pyodbc:///?odbc_connect='+ connection_string)
with engine.connect() as conn:
    df = pd.read_sql_query(query, con=conn)

#Store the data in Data directory
dest_path = '../Data/Raw_Balances_2026.xlsx'
df.to_excel(dest_path, index=False)

print('Execution Completed!')
```
