```python
from datetime import datetime
import dateutil

str1='20201031'
str2='20251031'

start_date = datetime.strptime(str1, '%Y%m%d')
end_date = datetime.strptime(str2, '%Y%m%d')

(end_date-start_date)/30
```

```python
diff = dateutil.relativedelta.relativedelta(end_date, start_date)
print(diff)
month_diff = diff.years * 12 + diff.months

print(str(month_diff) + ' months')
```
