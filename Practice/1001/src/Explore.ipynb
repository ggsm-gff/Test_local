```python
import sqlite3
import csv


# Functions
def get_session() -> sqlite3.Connection| sqlite3.Cursor:
    '''
    Opens a connection to a predefined database and return the Connection and Cursor Objects
    '''
    con = sqlite3.connect('../data/local.db')
    cur = con.cursor()
    return con, cur

def close_session(con:sqlite3.Connection , cur:sqlite3.Cursor):
    '''
    Takes the connection and cursor arguments and close the connection to the DB.  
    ''' 
    con.close()
    cur.close()

# DDL
def set_ddl_table(cur:sqlite3.Cursor):
    '''
    Tackes the cursor argumento to execute a ddl sentence.
    Creates a predefined sql table.
    '''
    # In sqlite the varchar type is considered TEXT which is legacy behavior
    
    ddl_sentence = """
    CREATE TABLE IF NOT EXISTS RAW_TBL_RESTAURANTS(
        ID INT,
        NAME TEXT,
        ADDR TEXT,
        CITY TEXT,
        PHONE TEXT,
        TYPE TEXT
    );
    """
    
    cur.execute(ddl_sentence)

con, cur = get_session()

set_ddl_table(cur)
```

```python
line_content = []
with open('../data/restaurants_L2_dirty.csv','r') as csv_file:
    content = csv.reader(csv_file, delimiter=',')
    for row in content:
        line_content.append(row)


cur.executemany("insert into raw_tbl_restaurants values(?,?,?,?,?,?)",line_content[1:])
con.commit()
```

```python

try:
    sql_retrieved = cur.execute("select * from raw_tbl_restaurants")
except Exception as e:
    print(e)
```
